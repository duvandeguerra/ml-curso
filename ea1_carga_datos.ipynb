# EA1 – Carga y creación de dataset

Flujo: cargar el dataset Iris desde Hugging Face con `HFDatasetLoader`,
aplicar transformaciones al DataFrame y generar un resumen en `.txt`
con `DatasetSummarizer`.

```python
from hf_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

## 1. Carga del dataset

```python
loader = HFDatasetLoader("scikit-learn/iris", split="train")
df = loader.load()

print("Dimensiones originales:", df.shape)
df.head()
```

## 2. Transformaciones

Se aplican cuatro transformaciones al DataFrame original:
1. Eliminar la columna `Id`, que solo numera filas y no aporta información.
2. Renombrar las columnas a español.
3. Crear la columna `area_petalo` (largo × ancho del pétalo).
4. Filtrar las flores con largo de sépalo mayor a 5 cm.

```python
df_t = df.drop(columns=["Id"])
df_t.head()
```

```python
df_t = df_t.rename(columns={
    "SepalLengthCm": "largo_sepalo",
    "SepalWidthCm": "ancho_sepalo",
    "PetalLengthCm": "largo_petalo",
    "PetalWidthCm": "ancho_petalo",
    "Species": "especie",
})
df_t.head()
```

```python
df_t["area_petalo"] = df_t["largo_petalo"] * df_t["ancho_petalo"]
df_t.head()
```

```python
df_t = df_t[df_t["largo_sepalo"] > 5.0]
print("Dimensiones después del filtro:", df_t.shape)
df_t.head()
```

```python
print("Original:     ", df.shape, list(df.columns))
print("Transformado: ", df_t.shape, list(df_t.columns))
```

## 3. Resumen del DataFrame transformado

```python
summarizer = DatasetSummarizer(df_t, "Iris transformado")
ruta = summarizer.save("resumen_iris.txt")

print(f"Resumen guardado en: {ruta}\n")
print(summarizer.build_summary())
```

## Conclusión

Se cargó el dataset Iris (150 filas, 6 columnas) mediante `HFDatasetLoader`.
Se eliminó la columna `Id`, se renombraron las columnas a español, se creó
`area_petalo` y se filtraron las flores con sépalo mayor a 5 cm.
El resumen de `resumen_iris.txt` corresponde al DataFrame transformado.
